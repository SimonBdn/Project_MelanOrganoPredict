```python
import polars as pl
```

```python
df_GDSC1 = pl.DataFrame(pl.read_excel("data/GDSC1.xlsx"))
df_GDSC1
```

```python
df_GDSC2 = pl.DataFrame(pl.read_excel("data/GDSC2.xlsx"))
df_GDSC2
```

```python
# 2. Fusion et suppression des doublons en utilisant la colonne "DATASET"
df_GDSC = (
    pl.concat([df_GDSC1, df_GDSC2], how="diagonal")
    .sort("DATASET", descending=True) # "GDSC2" passera toujours au-dessus de "GDSC1"
    .unique(subset=["CELL_LINE_NAME", "DRUG_NAME"], keep="first")
)

# 3. Affichage
print(f"Lignes GDSC1 : {df_GDSC1.height}")
print(f"Lignes GDSC2 : {df_GDSC2.height}")
print(f"Lignes du Master Dataset fusionné et nettoyé : {df_GDSC.height}")
```
